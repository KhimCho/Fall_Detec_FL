```python
import torch
from ultralytics import YOLO
```

```python
def main():
    data_file = '/kaggle/input/mul-fll-cam-dts/data_kg.yaml'  
    weights_file = 'yolov10n.pt'  

    img_size = (720, 480)  
    batch_size = 8  
    epochs = 100  

    model = YOLO(weights_file)

    results = model.train(
        data=data_file,
        imgsz=img_size,
        batch=batch_size,
        epochs=epochs,
        device='0', 
        # hyp='hyp.scratch.yaml', 
        lr0= 0.01,
        lrf= 0.1,
        momentum= 0.937,
        weight_decay= 0.0005,
        name='my_yolov10',  
        cache=True
    )

    print("Huấn luyện đã xong.")

if __name__ == '__main__':
    main()
```
